```python
import pandas as pd
import matplotlib.pyplot as plt
from datetime import datetime
```

```python
#TÍTULO DO SISTEMA
print("="*60)
print("CHARGEGRID INTELLIGENCE - GOODWE")
print("PROVA DE CONCEITO - SPRINT 2")
print("="*60)
```

```python
#VEÍCULOS CONECTADOS
veiculos = pd.DataFrame({
    "Veiculo":["EV01","EV02","EV03","EV04"],
    "Bateria (%)":[20,45,80,30]
})

print("VEÍCULOS CONECTADOS")
display(veiculos)
```

```python
#GERENCIAMENTO INTELIGENTE DE DEMANDA
energia_total = 100

prioridades = []

for bateria in veiculos["Bateria (%)"]:

    if bateria <= 30:
        prioridades.append(4)

    elif bateria <= 50:
        prioridades.append(3)

    elif bateria <= 70:
        prioridades.append(2)

    else:
        prioridades.append(1)

veiculos["Prioridade"] = prioridades

soma_prioridades = sum(prioridades)

potencias = []

for p in prioridades:
    potencia = round(
        (p/soma_prioridades) * energia_total,
        2
    )

    potencias.append(potencia)

veiculos["Potencia Recebida (kW)"] = potencias

print("DISTRIBUIÇÃO INTELIGENTE")
display(veiculos)
```

```python
#IA DE PREVISÃO DE DEMANDA
hora = datetime.now().hour

if 18 <= hora <= 22:
    previsao = "ALTA DEMANDA"

elif 12 <= hora < 18:
    previsao = "MEDIA DEMANDA"

else:
    previsao = "BAIXA DEMANDA"

print("PREVISÃO DA IA")
print("Status:", previsao)

if previsao == "ALTA DEMANDA":
    print("Ação: Utilizar baterias e reduzir cargas secundárias")

elif previsao == "MEDIA DEMANDA":
    print("Ação: Operação normal")

else:
    print("Ação: Priorizar energia solar")
```

```python
#ENERGIA SOLAR E SUSTENTABILIDADE
energia_solar = 40
energia_rede = 60

fontes = {
    "Energia Solar": energia_solar,
    "Rede Elétrica": energia_rede
}

print("FONTES DE ENERGIA")

for fonte, valor in fontes.items():
    print(f"{fonte}: {valor} kW")
```

```python
#GRÁFICO DE SUSTENTABILIDADE
plt.figure(figsize=(6,6))

plt.pie(
    fontes.values(),
    labels=fontes.keys(),
    autopct="%1.1f%%"
)

plt.title("Origem da Energia Utilizada")

plt.show()
```

```python
#TARIFAÇÃO AUTOMÁTICA
tarifa_kwh = 0.85

veiculos["Consumo (kWh)"] = (
    veiculos["Potencia Recebida (kW)"] * 0.5
)

veiculos["Valor (R$)"] = (
    veiculos["Consumo (kWh)"] * tarifa_kwh
).round(2)

print("TARIFAÇÃO AUTOMÁTICA")

display(
    veiculos[
        [
            "Veiculo",
            "Consumo (kWh)",
            "Valor (R$)"
        ]
    ]
)
```

```python
#DASHBOARD EXECUTIVO
total_energia = veiculos["Potencia Recebida (kW)"].sum()

total_faturado = veiculos["Valor (R$)"].sum()

veiculos_conectados = len(veiculos)

print("DASHBOARD EXECUTIVO")
print("-"*40)

print("Veículos conectados:", veiculos_conectados)

print("Energia distribuída:", total_energia,"kW")

print("Faturamento estimado: R$", round(total_faturado,2))

print("Previsão IA:", previsao)

print("-"*40)
```

```python
#GRÁFICO DE DISTRIBUIÇÃO DE ENERGIA
plt.figure(figsize=(8,4))

plt.bar(
    veiculos["Veiculo"],
    veiculos["Potencia Recebida (kW)"]
)

plt.title(
    "Distribuição Inteligente de Energia"
)

plt.ylabel("kW")

plt.show()
```

```python
#GRÁFICO DE COBRANÇA
plt.figure(figsize=(8,4))

plt.bar(
    veiculos["Veiculo"],
    veiculos["Valor (R$)"]
)

plt.title(
    "Tarifação por Veículo"
)

plt.ylabel("R$")

plt.show()
```
